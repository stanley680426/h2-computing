**Task 2.2**

Write a function `task2_2(list_of_integers)` that:

- takes a list of integers, `list_of_integers`
- sorts them into ascending order using merge sort
- returns the sorted list. **[7]**

Use the list `[56,25,4,98,0,18,4,5,7,0]` to test your function.

For example, the condition

`task2_2([56,25,4,98,0,18,4,5,7,0]) == [0,0,4,4,5,7,18,25,56,98]`

should return `True`. **[2]**


````python
# Task 2.2
def merge(left,right):
  merged = []
  i = j = 0
  while i < len(left) and j < len(right):
    if left[i] < right[j]:
      merged.append(left[i])
      i += 1
    else:
      merged.append(right[j])
      j += 1
  merged.extend(left[i:])
  merged.extend(right[j:])
  return merged

def task2_2(list_of_integers):
  if len(list_of_integers) <= 1:
    return list_of_integers

  mid = len(list_of_integers) // 2
  left = list_of_integers[:mid]
  right = list_of_integers[mid:]

  left_sorted = task2_2(left)
  right_sorted = task2_2(right)

  return merge(left_sorted,right_sorted)


result = task2_2([56,25,4,98,0,18,4,5,7,0])
print(result == [0,0,4,4,5,7,18,25,56,98] )
````

````python
# Task 2.3
def task2_3(filename_in,filename_out):
  with open(filename_in,'r') as file:#close
    list_of_integers = [line.strip() for line in file]

  sorted_integers = task2_2(list_of_integers)

  with open(filename_out,'w') as file:#close
    file.write(sorted_integers + '\n')

task2_3(
    "randomnumbers_<your name>_<centre number>_<index number>.txt",
    "sortednumbers_<your name>_<centre number>_<index number>.txt"
)
````
