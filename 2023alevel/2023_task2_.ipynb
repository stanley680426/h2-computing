````python
#Task 2.4
class Queue:
  def __init__(self):
    self.queueList = [None, None, None, None]
    self.headPointer = 0
    self.tailPointer = 0
    self.numberItems = 0

  def enqueue(self, item):
    if self.numberItems == len(self.queueList):
      return False
    self.queueList[self.tailPointer] = item
    self.tailPointer = (self.tailPointer + 1) % len(self.queueList)
    self.numberItems = self.numberItems + 1
    return True

  def dequeue(self):
    if self.numberItems == 0:
      return False
    item = self.queueList[self.headPointer]
    self.queueList[self.headPointer] = None
    self.headPointer = (self.headPointer + 1) % len(self.queueList)
    self.numberItems = self.numberItems - 1
    return item

````
